```python
import requests
import pandas as pd
```

```python
url = "https://api.worldbank.org/v2/country/SN;NG;FR;US;CN/indicator/IT.NET.USER.ZS?format=json&per_page=500"
reponse = requests.get(url)
data = reponse.json()
print(data)
```

```python
df = data[1]
dt = pd.DataFrame(df)
dt.head()
```

```python
dt['country'] = dt['country'].apply(lambda country: country['value'])

```

```python
dt['indicator'] = dt['indicator'].apply(lambda indicator : indicator['value'])
```

```python
dt['date']= dt['date'].astype(int)
```

```python
import seaborn as sn
sn.lineplot(data=dt,x='date' , y='value',hue='country')
```

```python

url = "https://api.worldbank.org/v2/country/SN;NG;FR;US;CN/indicator/NY.GDP.PCAP.CD?format=json&per_page=500"
reponse1 = requests.get(url)
data1 = reponse1.json()
print(data1)
```

```python
df1 = data1[1]
dt1 = pd.DataFrame(df1)
dt1.head()
```

```python
dt1['country'] = dt1['country'].apply(lambda country: country['value'])
dt1['indicator'] = dt1['indicator'].apply(lambda indicator : indicator['value'])
```

```python
dt1.head()
```

```python
dt1['date'] = dt1['date'].astype(int)
```

```python
fusion = dt.merge(dt1,on=['country','date'],how='inner')
print(fusion)
```

```python
len(fusion)
```

```python
fusion.columns
```

```python
fusion = fusion.rename(columns={'value_x': 'internet_pct','value_y':'pib_par_habitant'})
```

```python
corr = ['internet_pct','pib_par_habitant']
cor = fusion[corr].corr()
print(cor)
```

# Mini-rapport de conclusions — Projet Fracture numérique

## Contexte
Étude de la fracture numérique à travers 5 pays (Sénégal, Nigeria, France, États-Unis, Chine), en croisant le taux d'utilisateurs d'Internet (% de la population) avec le PIB par habitant, via l'API de la Banque mondiale.

## Méthodologie
1. Récupération des données via l'API World Bank : indicateur `IT.NET.USER.ZS` (accès Internet) pour SN, NG, FR, US, CN sur 1960-2025.
2. Nettoyage : extraction des champs imbriqués (`country`, `indicator`), conversion de `date` en entier.
3. Visualisation de l'évolution du taux d'accès à Internet par pays (courbe Seaborn, `lineplot` avec `hue='country'`).
4. Récupération d'un second indicateur, le PIB par habitant (`NY.GDP.PCAP.CD`), avec le même nettoyage.
5. Fusion des deux jeux de données sur `country` et `date` (`merge how='inner'`), puis calcul de la corrélation entre `internet_pct` et `pib_par_habitant`.

## Résultats clés
- **Évolution par pays** : le taux d'accès à Internet progresse fortement pour tous les pays depuis les années 2000, mais avec des trajectoires très différentes. Les États-Unis et la France atteignent des niveaux élevés (85-90 %) dès 2020, tandis que la Chine suit une progression plus graduelle sur la même période (70-76 % entre 2020 et 2022) avant de franchir ce même seuil plus tardivement, en 2023-2024. Le Sénégal et le Nigeria, partis de niveaux quasi nuls, restent nettement en retrait (autour de 40-60 % en 2023-2024).
- **Fusion des données** : la jointure interne sur pays et année conserve les 330 lignes des deux tables (5 pays × 66 années), montrant une couverture temporelle identique pour les deux indicateurs.
- **Corrélation avec le PIB** : corrélation positive forte (**0,72**) entre le taux d'accès à Internet et le PIB par habitant, confirmant que la richesse économique d'un pays est étroitement liée à son niveau de connectivité numérique.

## Conclusion
Les résultats confirment l'existence d'une fracture numérique corrélée au niveau de développement économique : plus le PIB par habitant est élevé, plus l'accès à Internet est répandu (corrélation de 0,72). Le Sénégal et le Nigeria illustrent ce retard, avec une croissance de l'accès à Internet plus tardive et un plafond encore nettement inférieur à celui des pays plus riches (France, États-Unis, Chine) — la Chine elle-même n'ayant rattrapé le niveau américain et français que très récemment (2023-2024). Cette corrélation ne prouve toutefois pas un lien de causalité direct : d'autres facteurs (infrastructures, politiques publiques, coût des équipements) jouent probablement un rôle dans cette dynamique.
