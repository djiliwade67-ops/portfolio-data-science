```python
import requests 
from bs4 import BeautifulSoup
```

```python
result = requests.get("https://keur-immo.com/senegal/")
recup = result.text
data = BeautifulSoup(recup, "html.parser")
```

```python
annonces = data.find_all("article", class_="g5core__gutter-item")
len(annonces) 
```

```python
text = annonces[0].find("h3", class_="g5ere__loop-property-title").text
print(text)
```

```python
prix = annonces[0].find("span", class_="g5ere__lpp-price").text
print(prix)
```

```python
chambre = annonces[0].find("span", class_="g5ere__property-bedrooms").text
print(chambre)
bain = annonces[0].find("span", class_="g5ere__loop-property-bathrooms").text
print(bain)
surface = annonces[0].find("span", class_="g5ere__loop-property-size").text
print(surface)
```
